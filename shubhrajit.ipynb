Print Mean AQI, Max PM2.5, Min PM10 for each city (using numpy)


```python
for i in df.City.unique():
    print(i)
    print('Mean AQI: \t', np.mean(df[df.City == i].AQI))
    print('Max PM2.5: \t', np.max(df[df.City == i]['PM2.5']))
    print('Min PM10: \t', np.min(df[df.City == i].PM10))
```

Group the df by City and find

    avg AQI
    max PM2.5
    min PM10

display as dictionary


```python
ans = df.groupby('City').agg({'AQI': 'mean', 'PM2.5': 'max', 'PM10': 'min'})
ans
```

```python
ans = ans.to_dict(orient='index')
ans
```

```python
ans = {k: [*v.values()] for k, v in ans.items()}
ans
```
